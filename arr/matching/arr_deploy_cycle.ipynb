# Post-Deployment Changes
The posted request form note has some important information for programatically setting the submission and post-submission fields. The URL for the request form `https://OpenReview.net/forum?id=<FORUM_ID>` contains the forum ID. You can get the request number by going to `https://api.OpenReview.net/notes?id=<FORUM_ID>` and finding the top-level `number` field.

Also make sure to update the `month_name` and `year` variables

````python
request_number = None
forum_id = None
copy_fields = ['title', 'Official Venue Name', 'Abbreviated Venue Name', 'Official Website URL', 'program_chair_emails', 'contact_email', 'publication_chairs', 'ethics_chairs_and_reviewers', 'Submission Start Date', 'Submission Deadline', 'Venue Start Date', 'submission_reviewer_assignment', 'Expected Submissions', 'use_recruitment_template', 'Additional Submission Options', 'remove_submission_options', 'submission_email', 'homepage_override']
program_chairs_id = f'aclweb.org/ACL/ARR/{year}/{month_name}/Program_Chairs'
````

````python
from openreview.stages.arr_content import arr_submission_content

forum_note = client_v1.get_note(forum_id)
original_content = {}
for key, value in forum_note.content.items():
    if key in copy_fields:
        original_content[key] = value
original_content['Additional Submission Options'] = arr_submission_content
original_content['homepage_override'] = {
  "title": f"ACL Rolling Review - {month_name} {year}",
  "deadline": f"Submission deadline on {month_name} 16 @ 23:59pm AOE",
  "date": month_name,
  "website": "http://aclrollingreview.org/",
  "instructions": "\u003Ch2\u003E For Authors\u003C/h2\u003E\u003Cp class=\"dark\"\u003EVisit the webpage of the official \u003Ca href=\"https://aclrollingreview.org/cfp\" target=\"_blank\"\u003Ecall for papers\u003C/a\u003E and please read the \u003Ca href=\"https://aclrollingreview.org/authors\" target=\"_blank\"\u003Eauthor information page\u003C/a\u003E, as well as the \u003Ca href=\"https://aclrollingreview.org/authorchecklist\" target=\"_blank\"\u003Eauthor checklist\u003C/a\u003E for submissions.\u003C/p\u003E\u003Ch2\u003E For Reviewers \u003C/h2\u003E\u003Cp class=\"dark\"\u003EPlease read the \u003Ca href=\"https://aclrollingreview.org/reviewing#reviewers\" target=\"_blank\"\u003Ereviewer information page\u003C/a\u003E.\u003C/p\u003E\u003Ch2\u003E For Action Editors \u003C/h2\u003E\u003Cp class=\"dark\"\u003EAlso please read the \u003Ca href=\"https://aclrollingreview.org/reviewing\" target=\"_blank\"\u003Ereviewer information page\u003C/a\u003E including information for action editors.\u003C/p\u003E\u003Cp class=\"dark\"\u003EPlease be aware that (for now) the action editors are called area chairs on OpenReview.\u003C/p\u003E\u003Ch2\u003E \u003Ca href=\"https://aclrollingreview.org/dates\" target=\"_blank\"\u003EReviewing Schedule\u003C/a\u003E \u003C/h2\u003E",
  "contact": "support@aclrollingreview.org"
}

client_v1.post_note(openreview.Note(
    invitation=f'OpenReview.net/Support/-/Request{request_number}/Revision',
    forum=forum_id,
    readers=[program_chairs_id, 'OpenReview.net/Support'],
    referent=forum_id,
    replyto=forum_id,
    signatures=[client.profile.id],
    writers=[],
    content=original_content
))
````

````python
from openreview.stages.arr_content import hide_fields

client_v1.post_note(openreview.Note(
    invitation=f'OpenReview.net/Support/-/Request{request_number}/Post_Submission',
    forum=forum_id,
    readers=[program_chairs_id, 'OpenReview.net/Support'],
    referent=forum_id,
    replyto=forum_id,
    signatures=[client.profile.id],
    writers=[],
    content= {
        'force': 'Yes',
        'submission_readers': 'Assigned program committee (assigned reviewers, assigned area chairs, assigned senior area chairs if applicable)',
        'hide_fields': hide_fields
    },
))
````

# Optional: Copy Members Only
Use this cell if tracks differ between cycles - which will break the typical copying process between venues

````python
current_venue = f"aclweb.org/ACL/ARR/{year}/{month_name}"
previous_venue = "aclweb.org/ACL/ARR/2026/<PREVIOUS_MONTH>"
print(current_venue)
roles = [
    'Area_Chairs',
    'Reviewers',
    'Ethics_Reviewers',
    'Ethics_Chairs'
]
for role in roles:
    prev_group = client.get_group(f"{previous_venue}/{role}")
    print(f"found {len(prev_group.members)} {role} in {previous_venue}")
    current_group = client.get_group(f"{current_venue}/{role}")
    client.add_members_to_group(
        group=current_group.id,
        members=prev_group.members
    )
    print(f"added {len(prev_group.members)} {role} to {current_group.id}")
    
````
