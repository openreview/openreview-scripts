# Post Submission Fields

```python
request_number = None
forum_id = ''
copy_fields = [
    'title',
    'Official Venue Name',
    'Abbreviated Venue Name',
    'Official Website URL',
    'program_chair_emails',
    'contact_email',
    'publication_chairs',
    'ethics_chairs_and_reviewers',
    'Submission Start Date',
    'Submission Deadline',
    'Venue Start Date',
    'submission_reviewer_assignment',
    'Expected Submissions',
    'use_recruitment_template',
    'Additional Submission Options',
    'remove_submission_options',
    'submission_email',
    'homepage_override'
]
program_chairs_id = 'aclweb.org/ACL/ARR/YEAR/MONTH/Program_Chairs'
month_name = 'MONTH'
year = 'YEAR'
```

```python
from openreview.stages.arr_content import arr_submission_content

forum_note = client_v1.get_note(forum_id)
original_content = {}
for key, value in forum_note.content.items():
    if key in copy_fields:
        original_content[key] = value
original_content['Additional Submission Options'] = arr_submission_content
original_content['homepage_override'] = {
  "title": f"ACL Rolling Review - {month_name} {year}",
  "deadline": f"Submission deadline on {month_name} 16 @ 23:59pm AOE",
  "date": month_name,
  "website": "http://aclrollingreview.org/",
  "instructions": "\u003Ch2\u003E For Authors\u003C/h2\u003E\u003Cp class=\"dark\"\u003EVisit the webpage of the official \u003Ca href=\"https://aclrollingreview.org/cfp\" target=\"_blank\"\u003Ecall for papers\u003C/a\u003E and please read the \u003Ca href=\"https://aclrollingreview.org/authors\" target=\"_blank\"\u003Eauthor information page\u003C/a\u003E, as well as the \u003Ca href=\"https://aclrollingreview.org/authorchecklist\" target=\"_blank\"\u003Eauthor checklist\u003C/a\u003E for submissions.\u003C/p\u003E\u003Ch2\u003E For Reviewers \u003C/h2\u003E\u003Cp class=\"dark\"\u003EPlease read the \u003Ca href=\"https://aclrollingreview.org/reviewing#reviewers\" target=\"_blank\"\u003Ereviewer information page\u003C/a\u003E.\u003C/p\u003E\u003Ch2\u003E For Action Editors \u003C/h2\u003E\u003Cp class=\"dark\"\u003EAlso please read the \u003Ca href=\"https://aclrollingreview.org/reviewing\" target=\"_blank\"\u003Ereviewer information page\u003C/a\u003E including information for action editors.\u003C/p\u003E\u003Cp class=\"dark\"\u003EPlease be aware that (for now) the action editors are called area chairs on OpenReview.\u003C/p\u003E\u003Ch2\u003E \u003Ca href=\"https://aclrollingreview.org/dates\" target=\"_blank\"\u003EReviewing Schedule\u003C/a\u003E \u003C/h2\u003E",
  "contact": "support@aclrollingreview.org"
}

client_v1.post_note(openreview.Note(
    invitation=f'OpenReview.net/Support/-/Request{request_number}/Revision',
    forum=forum_id,
    readers=[program_chairs_id, 'OpenReview.net/Support'],
    referent=forum_id,
    replyto=forum_id,
    signatures=['~Super_User1'],
    writers=[],
    content=original_content
))
```
